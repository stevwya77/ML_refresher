```python
# Q7 Sum of Weights
```

```python
selected = df[df["origin"] == 'Asia'][["vehicle_weight", "model_year"]].head(7)
x = np.array(selected)
x
```

```python
XTX = np.dot(x.T, x)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = np.dot(np.dot(XTX_inv, x.T), y)
w
```
